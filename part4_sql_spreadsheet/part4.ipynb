```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_FOLDER = Path(
    "/content/drive/MyDrive/TrendPulse/part2_numpy_pandas/data"
)

file_path = DATA_FOLDER / "cleaned_trendpulse.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
# 1. Create Categories table
categories = (
    df[["category_id", "category_name"]]
    .drop_duplicates()
    .sort_values("category_id")
    .reset_index(drop=True)
)

# 2. Create Videos table
video_columns = [
    "video_id",
    "title",
    "channel_title",
    "category_id",
    "publish_time",
    "tags",
    "thumbnail_link",
    "description",
    "comments_disabled",
    "ratings_disabled",
    "video_error_or_removed"
]

videos = (
    df[video_columns]
    .drop_duplicates(subset=["video_id"])
    .reset_index(drop=True)
)

# 3. Create Trending Records table
trending_columns = [
    "video_id",
    "trending_date",
    "country",
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

trending_records = (
    df[trending_columns]
    .drop_duplicates(
        subset=["video_id", "trending_date", "country"]
    )
    .reset_index(drop=True)
)

# Display table sizes
print("========== TABLE SUMMARY ==========")

print("Categories:", categories.shape)
print("Videos:", videos.shape)
print("Trending Records:", trending_records.shape)

print("\n========== CATEGORIES ==========")
display(categories.head())

print("\n========== VIDEOS ==========")
display(videos.head())

print("\n========== TRENDING RECORDS ==========")
display(trending_records.head())
```

```python
from pathlib import Path

# Create Part 4 data folder
OUTPUT_FOLDER = Path(
    "/content/drive/MyDrive/TrendPulse/part4_sql_spreadsheet/data"
)

OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

# Save the three datasets
categories.to_csv(OUTPUT_FOLDER / "categories.csv", index=False)

videos.to_csv(OUTPUT_FOLDER / "videos.csv", index=False)

trending_records.to_csv(
    OUTPUT_FOLDER / "trending_records.csv",
    index=False
)

print("All three CSV files saved successfully!")

print("\nFiles inside Part 4 data folder:")
for file in OUTPUT_FOLDER.iterdir():
    print(file.name)
```

```python
print("========== DATASET VALIDATION ==========")

print("\n1. CATEGORIES")
print("Rows:", len(categories))
print("Unique category IDs:", categories["category_id"].nunique())
print("Duplicate category IDs:", categories["category_id"].duplicated().sum())

print("\n2. VIDEOS")
print("Rows:", len(videos))
print("Unique video IDs:", videos["video_id"].nunique())
print("Duplicate video IDs:", videos["video_id"].duplicated().sum())

print("\n3. TRENDING RECORDS")
print("Rows:", len(trending_records))

duplicate_records = trending_records.duplicated(
    subset=["video_id", "trending_date", "country"]
).sum()

print("Duplicate video-date-country records:", duplicate_records)

# Check foreign-key relationships
print("\n4. RELATIONSHIP CHECKS")

missing_categories = (~videos["category_id"].isin(categories["category_id"])).sum()

missing_videos = (~trending_records["video_id"].isin(videos["video_id"])).sum()

print("Videos with missing category IDs:", missing_categories)
print("Trending records with missing video IDs:", missing_videos)
```

```python

import pandas as pd
import numpy as np

from google.colab import drive
from openpyxl import Workbook
from openpyxl.utils.dataframe import dataframe_to_rows
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.formatting.rule import Rule
from openpyxl.styles.differential import DifferentialStyle
from openpyxl.workbook.defined_name import DefinedName
from openpyxl.worksheet.table import Table, TableStyleInfo

# ==================================================
# 1. MOUNT GOOGLE DRIVE
# ==================================================

drive.mount('/content/drive')

folder = "/content/drive/MyDrive/TrendPulse/part4_sql_spreadsheet/data"

# ==================================================
# 2. LOAD DATASETS
# ==================================================

categories = pd.read_csv(f"{folder}/categories.csv")
videos = pd.read_csv(f"{folder}/videos.csv")
trending = pd.read_csv(f"{folder}/trending_records.csv")

print("Dataset shapes:")
print("Categories:", categories.shape)
print("Videos:", videos.shape)
print("Trending records:", trending.shape)

# ==================================================
# 3. CREATE WORKBOOK
# ==================================================

wb = Workbook()
wb.remove(wb.active)

datasets = {
    "Categories": categories,
    "Videos": videos,
    "Trending_Records": trending
}

# ==================================================
# 4. ADD DATA SHEETS
# ==================================================

for sheet_name, df in datasets.items():

    ws = wb.create_sheet(sheet_name)

    for row in dataframe_to_rows(
        df,
        index=False,
        header=True
    ):
        ws.append(row)

    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions

    # Header formatting
    for cell in ws[1]:
        cell.font = Font(
            bold=True,
            color="FFFFFF"
        )

        cell.fill = PatternFill(
            fill_type="solid",
            fgColor="1F4E78"
        )

        cell.alignment = Alignment(
            horizontal="center"
        )

    # Excel table
    table = Table(
        displayName=sheet_name + "Table",
        ref=ws.dimensions
    )

    table.tableStyleInfo = TableStyleInfo(
        name="TableStyleMedium2",
        showRowStripes=True,
        showColumnStripes=False
    )

    ws.add_table(table)

print("Data sheets created.")

# ==================================================
# 5. CREATE SUMMARY SHEET
# ==================================================

ws = wb.create_sheet("Summary")

summary_rows = [
    ["TrendPulse - Excel Analysis"],
    ["Metric", "Value"],
    ["Total Categories", "=COUNTA(Categories!B2:B100)"],
    ["Total Videos", "=COUNTA(Videos!A2:A20000)"],
    ["Total Trending Records", "=COUNTA(Trending_Records!A2:A50000)"],
    ["Total Views", "=SUM(Trending_Records!D2:D50000)"],
    ["Average Views", "=AVERAGE(Trending_Records!D2:D50000)"],
    ["US Record Count", '=COUNTIF(Trending_Records!C2:C50000,"US")'],
    ["GB Record Count", '=COUNTIF(Trending_Records!C2:C50000,"GB")'],
    ["IN Record Count", '=COUNTIF(Trending_Records!C2:C50000,"IN")'],
    ["GB Average Views", '=AVERAGEIF(Trending_Records!C2:C46001,"GB",Trending_Records!D2:D46001)'],
    ["US Average Views", '=AVERAGEIF(Trending_Records!C2:C46001,"US",Trending_Records!D2:D46001)'],
    ["IN Average Views", '=AVERAGEIF(Trending_Records!C2:C46001,"IN",Trending_Records!D2:D46001)']
]

for row in summary_rows:
    ws.append(row)

# ==================================================
# 6. XLOOKUP EXAMPLE
# ==================================================

ws["D1"] = "Category Lookup"
ws["D2"] = "Category ID"
ws["E2"] = "Category Name"

ws["D3"] = 10

ws["E3"] = (
    '=XLOOKUP(D3,Categories!A2:A100,'
    'Categories!B2:B100,"Not Found")'
)

# ==================================================
# 7. NAMED RANGE EXAMPLE
# ==================================================

ws["G1"] = "Named Range Example"
ws["G2"] = "Total Views"
ws["G3"] = "=SUM(TotalViews)"

wb.defined_names.add(
    DefinedName(
        "TotalViews",
        attr_text="'Trending_Records'!$D$2:$D$46001"
    )
)

# ==================================================
# 8. CATEGORY AND COUNTRY PIVOT SUMMARY
# ==================================================

category_country = (
    trending
    .merge(
        videos[["video_id", "category_id"]],
        on="video_id"
    )
    .merge(
        categories,
        on="category_id"
    )
)

pivot = (
    category_country
    .pivot_table(
        index="category_name",
        columns="country",
        values="views",
        aggfunc="mean"
    )
    .round(2)
    .reset_index()
)

pivot_ws = wb.create_sheet("Category_Pivot")

for row in dataframe_to_rows(
    pivot,
    index=False,
    header=True
):
    pivot_ws.append(row)

pivot_ws.freeze_panes = "A2"

print("Category and country summary created.")

# ==================================================
# 9. FORMAT SUMMARY AND PIVOT SHEETS
# ==================================================

for sheet in [ws, pivot_ws]:

    for cell in sheet[1]:
        cell.font = Font(
            bold=True,
            color="FFFFFF"
        )

        cell.fill = PatternFill(
            fill_type="solid",
            fgColor="1F4E78"
        )

    for column_cells in sheet.columns:

        max_length = max(
            len(str(cell.value or ""))
            for cell in column_cells
        )

        sheet.column_dimensions[
            column_cells[0].column_letter
        ].width = min(max_length + 3, 35)

# ==================================================
# 10. CONDITIONAL FORMATTING - TOP 10% VIEWS
# ==================================================

trending_ws = wb["Trending_Records"]

views_col = None

for cell in trending_ws[1]:

    if cell.value == "views":
        views_col = cell.column_letter
        break

if views_col:

    top_10_rule = Rule(
        type="top10",
        rank=10,
        percent=True,
        dxf=DifferentialStyle(
            fill=PatternFill(
                fill_type="solid",
                fgColor="FFD966"
            )
        )
    )

    trending_ws.conditional_formatting.add(
        f"{views_col}2:{views_col}46001",
        top_10_rule
    )

print("Top 10% conditional formatting added.")

# ==================================================
# 11. SAVE EXCEL WORKBOOK
# ==================================================

output_path = f"{folder}/TrendPulse_Analysis.xlsx"

wb.save(output_path)

print("\n===================================")
print("EXCEL WORKBOOK CREATED SUCCESSFULLY")
print("===================================")

print("Saved at:", output_path)

print("\nSheets created:")

for sheet in wb.sheetnames:
    print("-", sheet)
```

```python

import pandas as pd
from openpyxl import load_workbook
from openpyxl.utils.dataframe import dataframe_to_rows
from openpyxl.styles import Font, PatternFill
from openpyxl.worksheet.table import Table, TableStyleInfo

folder = "/content/drive/MyDrive/TrendPulse/part4_sql_spreadsheet/data"

# Join trending records with video and category information
pivot_source = (
    trending
    .merge(
        videos[["video_id", "category_id"]],
        on="video_id"
    )
    .merge(
        categories[["category_id", "category_name"]],
        on="category_id"
    )
)

# Keep only the columns needed for the PivotTable
pivot_source = pivot_source[
    ["video_id", "trending_date", "country",
     "category_name", "views"]
]

# Open existing workbook
file_path = f"{folder}/TrendPulse_Analysis.xlsx"
wb = load_workbook(file_path)

# Remove old sheet if rerunning
if "Pivot_Source" in wb.sheetnames:
    del wb["Pivot_Source"]

# Create source sheet
ws = wb.create_sheet("Pivot_Source")

for row in dataframe_to_rows(
    pivot_source,
    index=False,
    header=True
):
    ws.append(row)

ws.freeze_panes = "A2"

# Format headers
for cell in ws[1]:
    cell.font = Font(bold=True, color="FFFFFF")
    cell.fill = PatternFill("solid", fgColor="1F4E78")

# Create Excel table
table = Table(
    displayName="PivotSourceTable",
    ref=ws.dimensions
)

table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium2",
    showRowStripes=True
)

ws.add_table(table)

wb.save(file_path)

print("Pivot_Source sheet created!")
print("Rows:", len(pivot_source))
print("Columns:", list(pivot_source.columns))
```
