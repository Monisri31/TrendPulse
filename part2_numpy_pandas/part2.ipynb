```python
import numpy as np
import pandas as pd

# Create sample arrays
views = np.array([1000, 2500, 5000, 10000, 20000])
likes = np.array([100, 300, 600, 1200, 2500])
dislikes = np.array([10, 20, 50, 100, 200])
comments = np.array([5, 15, 30, 60, 100])

print("NumPy imported successfully!")
print("Views:", views)
print("Likes:", likes)
print("Dislikes:", dislikes)
print("Comments:", comments)
```

```python
total_ratings = likes + dislikes

like_ratio = np.divide(
    likes,
    total_ratings,
    out=np.zeros_like(likes, dtype=float),
    where=total_ratings != 0
)

print("Total ratings:", total_ratings)
print("Like ratio:", like_ratio)
```

```python
comment_rate = np.divide(
    comments,
    views,
    out=np.zeros_like(views, dtype=float),
    where=views != 0
)

print("Comment rate:", comment_rate)
```

```python
min_views = np.min(views)
max_views = np.max(views)

normalized_views = (
    (views - min_views) / (max_views - min_views)
)

print("Minimum views:", min_views)
print("Maximum views:", max_views)
print("Normalized views:", normalized_views)
```

```python
log_views = np.log1p(views)

print("Log transformed views:", log_views)
```

```python
view_threshold = np.percentile(views, 90)

top_10_percent = views[views >= view_threshold]

print("90th percentile:", view_threshold)
print("Top 10% views:", top_10_percent)
```

```python
normalized_likes = (
    (likes - np.min(likes)) /
    (np.max(likes) - np.min(likes))
)

normalized_comments = (
    (comments - np.min(comments)) /
    (np.max(comments) - np.min(comments))
)

weighted_engagement = (
    0.5 * normalized_likes +
    0.3 * normalized_comments +
    0.2 * normalized_views
)

print("Weighted engagement:", weighted_engagement)
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

from google.colab import drive
drive.mount('/content/drive')

DATA_FOLDER = Path(
    "/content/drive/MyDrive/TrendPulse/part2_numpy_pandas/data"
)

us = pd.read_csv(DATA_FOLDER / "USvideos.csv")
gb = pd.read_csv(DATA_FOLDER / "GBvideos.csv")
india = pd.read_csv(DATA_FOLDER / "INvideos.csv")
categories = pd.read_csv(DATA_FOLDER / "categories.csv")

print("US:", us.shape)
print("GB:", gb.shape)
print("India:", india.shape)
print("Categories:", categories.shape)
```

```python
us["country"] = "US"
gb["country"] = "GB"
india["country"] = "IN"

all_videos = pd.concat(
    [us, gb, india],
    ignore_index=True
)

print("Combined shape:", all_videos.shape)
display(all_videos.head())
```

```python
print("Dataset shape:", all_videos.shape)

print("\nColumn names:")
print(all_videos.columns.tolist())

print("\nData types:")
print(all_videos.dtypes)

print("\nDataset information:")
all_videos.info()
```

```python
print("Missing values per column:")
print(all_videos.isnull().sum())
```

```python
print("Full duplicate rows:", all_videos.duplicated().sum())

print(
    "Duplicate video-date-country records:",
    all_videos.duplicated(
        subset=["video_id", "trending_date", "country"]
    ).sum()
)
```

```python
all_videos["trending_date"] = pd.to_datetime(
    all_videos["trending_date"],
    format="%y.%d.%m",
    errors="coerce"
)

all_videos["publish_time"] = pd.to_datetime(
    all_videos["publish_time"],
    errors="coerce",
    utc=True
)

print(all_videos[["trending_date", "publish_time"]].dtypes)

display(all_videos[["trending_date", "publish_time"]].head())
```

```python
for column in ["likes", "comment_count"]:
    all_videos[column] = all_videos[column].fillna(
        all_videos.groupby("country")[column].transform("median")
    )

all_videos["description"] = all_videos["description"].fillna("")

print("Missing values after cleaning:")
print(all_videos.isnull().sum())
```

```python
before = len(all_videos)

all_videos = all_videos.drop_duplicates(
    subset=["video_id", "trending_date", "country"]
)

after = len(all_videos)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)
```

```python
print("Final shape:", all_videos.shape)
print("Remaining missing values:", all_videos.isnull().sum().sum())
print("Remaining duplicates:", all_videos.duplicated().sum())
```

```python
# 1. Videos with more than 1 million views
viral_videos = all_videos[all_videos["views"] > 1000000]

print("Videos with more than 1 million views:", len(viral_videos))


# 2. Videos with more than 1 million views AND more than 50,000 likes
high_engagement = all_videos[
    (all_videos["views"] > 1000000) &
    (all_videos["likes"] > 50000)
]

print("High engagement videos:", len(high_engagement))


# 3. Videos from India
india_videos = all_videos[all_videos["country"] == "IN"]

print("India videos:", len(india_videos))


# 4. Select specific columns using loc
selected_videos = all_videos.loc[
    all_videos["country"] == "IN",
    ["title", "views", "likes", "country"]
]

print(selected_videos.head())


# 5. Select first 5 rows and first 4 columns using iloc
print(all_videos.iloc[:5, :4])


# 6. Sort videos by views in descending order
top_videos = all_videos.sort_values(
    by="views",
    ascending=False
)

print(top_videos[["title", "views", "country"]].head(10))
```

```python
# Check category dataset
print(categories.head())

# Merge category names into all_videos
all_videos = all_videos.merge(
    categories,
    on="category_id",
    how="left"
)

# Display results
print("Updated shape:", all_videos.shape)

print(all_videos[
    ["title", "category_id", "category_name", "country"]
].head(10))

# Check missing category names
print("Missing category names:", all_videos["category_name"].isna().sum())
```

```python
category_analysis = all_videos.groupby("category_name")["views"].mean()

print(category_analysis.sort_values(ascending=False))
```

```python
country_category = all_videos.groupby(
    ["country", "category_name"]
).agg(
    total_videos=("video_id", "count"),
    average_views=("views", "mean"),
    total_likes=("likes", "sum"),
    average_comments=("comment_count", "mean")
).reset_index()

print(country_category.head(15))
```

```python
category_counts = all_videos["category_name"].value_counts()

print(category_counts.head(10))
```

```python
category_percentages = (
    all_videos["category_name"].value_counts(normalize=True) * 100
)

print(category_percentages.round(2))
```

```python
# Extract publishing day and hour
all_videos["publish_day"] = all_videos["publish_time"].dt.day_name()
all_videos["publish_hour"] = all_videos["publish_time"].dt.hour

# Count records by publishing day
day_counts = all_videos["publish_day"].value_counts()

print("Publishing day counts:")
print(day_counts)

# Count records by publishing hour
hour_counts = all_videos["publish_hour"].value_counts().sort_index()

print("\nPublishing hour counts:")
print(hour_counts)

# Find the most common publishing day and hour
print("\nMost common publishing day:", day_counts.idxmax())
print("Most common publishing hour:", hour_counts.idxmax())
```

```python
# Find the first trending date for each video in each country
first_trending = all_videos.groupby(
    ["video_id", "country"]
)["trending_date"].min().reset_index()

first_trending.rename(
    columns={"trending_date": "first_trending_date"},
    inplace=True
)

# Get one publication time per video and country
publish_dates = all_videos.groupby(
    ["video_id", "country"]
)["publish_time"].min().reset_index()

# Merge publication date and first trending date
video_timing = first_trending.merge(
    publish_dates,
    on=["video_id", "country"],
    how="left"
)

# Calculate days taken to trend
video_timing["days_to_trend"] = (
    video_timing["first_trending_date"] -
    video_timing["publish_time"].dt.tz_localize(None).dt.normalize()
).dt.days

print(video_timing.head(10))

print("\nDays to trend statistics:")
print(video_timing["days_to_trend"].describe())
```

```python
trending_lifespan = all_videos.groupby(
    ["video_id", "country"]
).agg(
    first_trending_date=("trending_date", "min"),
    last_trending_date=("trending_date", "max"),
    trending_days=("trending_date", "nunique")
).reset_index()

print(trending_lifespan.head(10))

print("\nTrending lifespan statistics:")
print(trending_lifespan["trending_days"].describe())
```

```python
video_summary = all_videos.groupby(
    ["video_id", "country"]
).agg(
    title=("title", "first"),
    channel_title=("channel_title", "first"),
    category_name=("category_name", "first"),
    peak_views=("views", "max"),
    average_views=("views", "mean"),
    total_likes=("likes", "max"),
    total_comments=("comment_count", "max"),
    trending_days=("trending_date", "nunique")
).reset_index()

print("Video summary shape:", video_summary.shape)
print(video_summary.head())
```

```python
channel_summary = all_videos.groupby(
    ["channel_title", "country"]
).agg(
    total_trending_records=("video_id", "count"),
    unique_videos=("video_id", "nunique"),
    average_views=("views", "mean"),
    maximum_views=("views", "max"),
    total_likes=("likes", "sum")
).reset_index()

print("Channel summary shape:", channel_summary.shape)
print(channel_summary.head())
```

```python
us_summary = video_summary[video_summary["country"] == "US"]
gb_summary = video_summary[video_summary["country"] == "GB"]
in_summary = video_summary[video_summary["country"] == "IN"]

combined_summary = pd.concat(
    [us_summary, gb_summary, in_summary],
    ignore_index=True
)

print("Combined summary shape:", combined_summary.shape)
```

```python
final_video_summary = video_summary.merge(
    video_timing[["video_id", "country", "days_to_trend"]],
    on=["video_id", "country"],
    how="left"
)

print("Final video summary shape:", final_video_summary.shape)

print(final_video_summary[
    ["video_id", "country", "peak_views", "trending_days", "days_to_trend"]
].head(10))
```

```python
# Convert Pandas columns into NumPy arrays
views_array = all_videos["views"].to_numpy()
likes_array = all_videos["likes"].to_numpy()
dislikes_array = all_videos["dislikes"].to_numpy()
comments_array = all_videos["comment_count"].to_numpy()

# Check array properties
print("Views shape:", views_array.shape)
print("Views dtype:", views_array.dtype)

# Calculate like ratio safely
total_ratings = likes_array + dislikes_array

like_ratio = np.divide(
    likes_array,
    total_ratings,
    out=np.zeros_like(likes_array, dtype=float),
    where=total_ratings != 0
)

# Calculate comment rate safely
comment_rate = np.divide(
    comments_array,
    views_array,
    out=np.zeros_like(comments_array, dtype=float),
    where=views_array != 0
)

# Min-max normalization of views
min_views = views_array.min()
max_views = views_array.max()

normalized_views = (
    (views_array - min_views) / (max_views - min_views)
)

# Log transformation
log_views = np.log1p(views_array)

# Top 10% view threshold
view_threshold = np.percentile(views_array, 90)
top_10_percent = views_array[views_array >= view_threshold]

print("\nAverage like ratio:", round(like_ratio.mean(), 4))
print("Average comment rate:", round(comment_rate.mean(), 6))
print("Minimum views:", min_views)
print("Maximum views:", max_views)
print("90th percentile views:", view_threshold)
print("Videos in top 10%:", len(top_10_percent))

print("\nFirst 5 normalized views:", normalized_views[:5])
print("First 5 log views:", log_views[:5])
```

```python
# Save cleaned dataset
all_videos.to_csv(
    DATA_FOLDER / "cleaned_trendpulse.csv",
    index=False
)

# Save video summary
final_video_summary.to_csv(
    DATA_FOLDER / "video_summary.csv",
    index=False
)

# Save channel summary
channel_summary.to_csv(
    DATA_FOLDER / "channel_summary.csv",
    index=False
)

print("Part 2 output files saved successfully!")
```
