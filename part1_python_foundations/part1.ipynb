```python
# Fill missing likes and comments using each country's median
for column in ["likes", "comment_count"]:
    all_videos[column] = all_videos[column].fillna(
        all_videos.groupby("country")[column].transform("median")
    )

# Fill missing descriptions with empty text
all_videos["description"] = all_videos["description"].fillna("")

print("Missing values after cleaning:")
print(all_videos.isnull().sum())
```

```python
# Convert numeric columns
numeric_columns = [
    "category_id",
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

for column in numeric_columns:
    all_videos[column] = pd.to_numeric(
        all_videos[column], errors="coerce"
    )

# Convert likes and comments to integers
all_videos["likes"] = all_videos["likes"].round().astype(int)
all_videos["comment_count"] = all_videos["comment_count"].round().astype(int)

# Convert status columns to Boolean
boolean_columns = [
    "comments_disabled",
    "ratings_disabled",
    "video_error_or_removed"
]

for column in boolean_columns:
    all_videos[column] = all_videos[column].astype(bool)

print(all_videos.dtypes)
```

```python
import string

def title_features(title):

    words = title.split()

    return {
        "characters": len(title),
        "words": len(words),
        "uppercase": sum(1 for char in title if char.isupper()),
        "punctuation": sum(1 for char in title if char in string.punctuation),
        "average_word_length": (
            sum(len(word) for word in words) / len(words)
            if words else 0
        )
    }

# Test the function
sample_title = "Honest Football Prank LEAKED!"

print(title_features(sample_title))
```

```python
title_results = all_videos["title"].apply(title_features)

title_features_df = pd.DataFrame(
    title_results.tolist()
)

display(title_features_df.head())
```

```python
def classify_video(record):

    views = record["views"]
    likes = record["likes"]
    dislikes = record["dislikes"]
    comments = record["comment_count"]

    # Check disabled ratings first
    if record["ratings_disabled"]:
        return "Unrated"

    # Calculate like ratio safely
    total_ratings = likes + dislikes

    like_ratio = (
        likes / total_ratings
        if total_ratings > 0
        else 0
    )

    # Viral Hit
    if views >= 1000000:
        return "Viral Hit"

    # Divisive
    elif total_ratings >= 100 and like_ratio < 0.70:
        return "Divisive"

    # Steady Performer
    elif views >= 100000 and like_ratio >= 0.70:
        return "Steady Performer"

    # Low Engagement
    else:
        return "Low Engagement"
```

```python
def load_country(path, country, max_rows=None):

    try:
        df = pd.read_csv(path, nrows=max_rows)

        # Add country information
        df["country"] = country

        # Required columns
        required_columns = [
            "video_id",
            "trending_date",
            "title",
            "channel_title",
            "category_id",
            "publish_time",
            "views",
            "likes",
            "dislikes",
            "comment_count"
        ]

        # Check required columns
        missing_columns = [
            col for col in required_columns
            if col not in df.columns
        ]

        if missing_columns:
            raise ValueError(
                f"Missing columns: {missing_columns}"
            )

        return df

    except Exception as e:
        print("Error loading dataset:", e)
        return None
```

```python

def classify_video(record):

    # Handle missing or invalid information
    views = record.get("views")
    likes = record.get("likes")
    dislikes = record.get("dislikes")

    if pd.isna(views) or pd.isna(likes) or pd.isna(dislikes):
        return "Unrated"

    # Disabled ratings or removed videos
    if record.get("ratings_disabled", False) or record.get("video_error_or_removed", False):
        return "Unrated"

    total_ratings = likes + dislikes

    like_ratio = (
        likes / total_ratings
        if total_ratings > 0
        else 0
    )

    if views >= 1000000:
        return "Viral Hit"

    elif total_ratings >= 100 and like_ratio < 0.70:
        return "Divisive"

    elif views >= 100000 and like_ratio >= 0.70:
        return "Steady Performer"

    else:
        return "Low Engagement"
```

```python
all_videos["video_classification"] = all_videos.apply(
    lambda row: classify_video(row),
    axis=1
)
```
