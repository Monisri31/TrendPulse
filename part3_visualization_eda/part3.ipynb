```python
# Create a copy for Part 3 analysis
df_eda = df.copy()

# 1. Handle missing descriptions
df_eda["description"] = df_eda["description"].fillna("")

# 2. Clean category names
df_eda["category_name"] = df_eda["category_name"].str.strip()

# 3. Convert date columns
df_eda["trending_date"] = pd.to_datetime(
    df_eda["trending_date"], errors="coerce"
)

df_eda["publish_time"] = pd.to_datetime(
    df_eda["publish_time"], errors="coerce", utc=True
)

# 4. Remove duplicate records if any
df_eda = df_eda.drop_duplicates()

# 5. Verify cleaning
print("Data cleaning completed!")

print("\nUpdated dataset shape:", df_eda.shape)

print("\nMissing values after cleaning:")
print(df_eda.isnull().sum())

print("\nTotal missing values:",
      df_eda.isnull().sum().sum())

print("\nDuplicate rows:",
      df_eda.duplicated().sum())

print("\nCategory names:")
print(sorted(df_eda["category_name"].unique()))
```

```python
# 1. Dataset dimensions
total_rows = len(df_eda)
total_columns = len(df_eda.columns)

# 2. Country distribution
country_summary = df_eda["country"].value_counts().reset_index()
country_summary.columns = ["Country", "Video Records"]

# 3. Category distribution
category_summary = (
    df_eda["category_name"]
    .value_counts()
    .reset_index()
)

category_summary.columns = ["Category", "Video Records"]

# 4. Numeric statistics
numeric_summary = df_eda[
    ["views", "likes", "dislikes", "comment_count"]
].describe().round(2)

# 5. Missing value percentage
missing_percentage = (
    df_eda.isnull().mean() * 100
).round(2)

# 6. Create profile summary
print("========== DATA PROFILE SUMMARY ==========")

print("\nTotal Rows:", total_rows)
print("Total Columns:", total_columns)

print("\nCountry Distribution:")
display(country_summary)

print("\nCategory Distribution:")
display(category_summary)

print("\nNumeric Statistics:")
display(numeric_summary)

print("\nMissing Value Percentage:")
display(missing_percentage)
```

```python
# Statistical Analysis: Views

views_mean = df_eda["views"].mean()
views_median = df_eda["views"].median()
views_mode = df_eda["views"].mode()[0]
views_skewness = df_eda["views"].skew()

print("========== VIEWS STATISTICAL ANALYSIS ==========")

print("Mean:", round(views_mean, 2))
print("Median:", round(views_median, 2))
print("Mode:", views_mode)
print("Skewness:", round(views_skewness, 2))
```

```python
# IQR Outlier Detection for Views

Q1 = df_eda["views"].quantile(0.25)
Q3 = df_eda["views"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

outliers = df_eda[
    (df_eda["views"] < lower_bound) |
    (df_eda["views"] > upper_bound)
]

print("========== IQR OUTLIER ANALYSIS ==========")

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))
print("Number of Outliers:", len(outliers))
print("Percentage of Outliers:", round(len(outliers) / len(df_eda) * 100, 2), "%")
```

```python
# Correlation Analysis

numeric_columns = [
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

correlation_matrix = df_eda[numeric_columns].corr()

print("========== CORRELATION MATRIX ==========")

display(correlation_matrix.round(3))

# Create Heatmap

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Video Engagement Metrics")

plt.tight_layout()
plt.show()
```

```python
# Video Performance Funnel Analysis

total_records = len(df_eda)

views_100k = (df_eda["views"] >= 100000).sum()

views_1m = (df_eda["views"] >= 1000000).sum()

high_engagement = (
    (df_eda["views"] >= 1000000) &
    (df_eda["likes"] >= 50000)
).sum()

funnel_data = pd.DataFrame({
    "Stage": [
        "All Video Records",
        "Views >= 100K",
        "Views >= 1M",
        "Views >= 1M and Likes >= 50K"
    ],
    "Records": [
        total_records,
        views_100k,
        views_1m,
        high_engagement
    ]
})

funnel_data["Percentage"] = (
    funnel_data["Records"] / total_records * 100
).round(2)

print("========== VIDEO PERFORMANCE FUNNEL ==========")

display(funnel_data)
```

```python
# Video Retention Analysis

video_duration = (
    df_eda.groupby(["video_id", "country"])["trending_date"]
    .nunique()
    .reset_index(name="trending_days")
)

total_videos = len(video_duration)

retention_days = [2, 3, 5, 7]

retention_results = []

for day in retention_days:

    retained_videos = (
        video_duration["trending_days"] >= day
    ).sum()

    retention_percentage = (
        retained_videos / total_videos * 100
    )

    retention_results.append({
        "Day": day,
        "Retained Videos": retained_videos,
        "Retention Percentage": round(retention_percentage, 2)
    })

retention_df = pd.DataFrame(retention_results)

print("========== VIDEO TREND RETENTION ==========")

print("Total Unique Video-Country Combinations:", total_videos)

display(retention_df)
```

```python
# Cohort Analysis: First Trending Week and Month

cohort_df = (
    df_eda.groupby(["video_id", "country"])["trending_date"]
    .min()
    .reset_index(name="first_trending_date")
)

# Extract first trending week and month

cohort_df["first_trending_week"] = (
    cohort_df["first_trending_date"]
    .dt.to_period("W")
    .astype(str)
)

cohort_df["first_trending_month"] = (
    cohort_df["first_trending_date"]
    .dt.to_period("M")
    .astype(str)
)

# Weekly cohort summary

weekly_cohort = (
    cohort_df.groupby("first_trending_week")
    .size()
    .reset_index(name="New Trending Videos")
)

# Monthly cohort summary

monthly_cohort = (
    cohort_df.groupby("first_trending_month")
    .size()
    .reset_index(name="New Trending Videos")
)

print("========== WEEKLY COHORT ANALYSIS ==========")
display(weekly_cohort)

print("========== MONTHLY COHORT ANALYSIS ==========")
display(monthly_cohort)
```

6. Content teams can examine category-level view counts, engagement patterns, and publication timing to identify promising content opportunities.
